# Load the training data

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

print('load_data...')
df = pd.read_csv('gesture_data.csv')
X = df[['vector_x','vector_y']]
y = df['label']
print('data successfully loaded')
```

# Process the data

```python
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)
print(f'class encoding:{encoder.classes_} ')
X_train,X_test,y_train,y_test = train_test_split(X,y_encoded,test_size=0.2,random_state=42)
```

```python
import keras
from keras import layers

model = keras.Sequential([
    layers.Input(shape=(2,)),
    layers.Dense(16,activation='relu'),
    layers.Dense(4,activation='softmax')

],)
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model.fit(X_train,y_train,epochs=100)
```

# Test the model

```python
import numpy as np

# test vector: strongly negative y should be "up"
test_vector = np.array([[0, -150]])

# probabilities for the 4 classes
prediction = model.predict(test_vector)

class_index = np.argmax(prediction)

# index back to the label
result = encoder.classes_[class_index]
print(f"The model predicts: {result}")
```

# Save the model

```python
import pickle
model.save('gesture_model.keras')
print('Keras model saved as "gesture_model.keras"')
# save the LabelEncoder
with open('label_encoder.pkl','wb')as f:
    pickle.dump(encoder,f)
print('LabelEncoder saved as "label_encoder.pkl"')
```
